# E-Commerce Product Compare Scraper — Analysis Notebook

A quick interactive walkthrough of the scraping pipeline, data cleaning, analysis and visualisation.

## 1. Scrape products

Provide one or more product URLs, or a category/search URL. The pipeline follows pagination automatically, cleans and validates the data, and saves CSV/Excel + charts.

```python
from scraper.pipeline import scrape_products

# Example: scrape the 'Travel' category from the Books to Scrape sandbox site.
result = scrape_products(
    urls=["https://books.toscrape.com/catalogue/category/books/travel_2/index.html"],
    max_pages=3,
    min_delay=1.0,
    do_analyze=True,
    do_visualize=True,
)

products = result["products"]
print(f"Scraped {len(products)} products")
for p in products:
    print(f"- {p.name} | {p.price} {p.currency} | rating={p.rating} | {p.category}")
```

## 2. Examine the DataFrame

```python
from scraper.storage import products_to_dataframe

df = products_to_dataframe(result["products"])
df.head()
```

## 3. Summary statistics

```python
from scraper.analysis import analyze_products
import json

stats = analyze_products(df)
print(json.dumps(
    {k: v for k, v in stats.items() if k not in ('top_rated', 'cheapest', 'most_expensive')},
    indent=2, default=str
))
```

## 4. Visualisations

```python
import matplotlib.pyplot as plt
from scraper.visualization import (
    plot_price_distribution,
    plot_rating_distribution,
    plot_price_vs_rating,
    plot_products_by_category,
)

plt.figure(figsize=(14, 10))
plt.subplot(2, 2, 1)
fig = plot_price_distribution(df)
ax = fig.axes[0]
plt.sca(ax)

plt.tight_layout()
plt.show()
```

## 5. Export to CSV and Excel

```python
from scraper.storage import save_csv, save_excel

save_csv(df, "output/products.csv")
save_excel(df, "output/products.xlsx")
print("Saved to output/ (see project folder)")
```
